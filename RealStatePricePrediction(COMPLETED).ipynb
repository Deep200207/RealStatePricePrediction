# **Data Cleaning And Preprocessing**

```python
df=df.drop(columns=['SUPER_SQFT','BUILTUP_SQFT','SUPERBUILTUP_SQFT','CARPET_SQFT'],axis=1)
```

```python
df['BEDROOM_NUM'] = df['BEDROOM_NUM'].fillna(df['BEDROOM_NUM'].median())
df['BALCONY_NUM'] = df['BALCONY_NUM'].fillna(df['BALCONY_NUM'].median())
```

```python
df['FLOOR_NUM']=pd.to_numeric(df['FLOOR_NUM'],errors='coerce')
df['FLOOR_NUM'].info()
```

```python
df['FLOOR_NUM']=df['FLOOR_NUM'].fillna(df['FLOOR_NUM'].median())
```

```python
df.isnull().sum()
```

```python
df['TOTAL_FLOOR']=df['TOTAL_FLOOR'].fillna(df['TOTAL_FLOOR'].median())
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
print(df['LOCALITY'].unique().size)
print(df['PROPERTY_TYPE'].unique())
df_selected=df
```

```python
df_selected['LOCATION_TARGET']=df_selected.groupby('LOCALITY')['PRICE_PER_UNIT_AREA'].transform('mean')
df_selected=df_selected.drop(columns=['LOCALITY'],axis=1)
```

```python
df_selected.head()
```

```python
df_selected['AREA']=df_selected['AREA'].str.replace("sq.ft.","",regex=False)
df_selected['AREA']=df_selected['AREA'].str.replace(",","",regex=False)
df_selected['AREA']=df_selected['AREA'].str.strip()
mask = df_selected['AREA'].str.contains('-')
df_selected.loc[mask, 'AREA'] = df_selected.loc[mask, 'AREA'].str.split('-').apply(lambda x: (float(x[0])+float(x[1]))/2)
df_selected['AREA']=pd.to_numeric(df_selected['AREA'])
```

```python
df_selected['TOTAL_FLOOR']=pd.to_numeric(df_selected['TOTAL_FLOOR'])
df_selected['FLOOR_NUM']=pd.to_numeric(df_selected['FLOOR_NUM'])
df_selected.info()
```

```python
df_selected = df_selected[df_selected['MAX_PRICE'] > 0]
df_selected = pd.get_dummies(df_selected, columns=['PROPERTY_TYPE'], drop_first=True)
df_selected.head()
```

```python
plt.figure(figsize=(10,5))
attributes=['PRICE_PER_UNIT_AREA','MAX_PRICE','BATHROOM_NUM','BEDROOM_NUM','AREA','LOCATION_TARGET']
sns.boxplot(data=df_selected[attributes],whis=1.5,width=0.7)
plt.xlabel("Attributes",fontweight='bold')
plt.ylabel("Ranges",fontweight='bold')
```

```python
# price_limit = df['MAX_PRICE'].quantile(0.95)
# ppu_limit=df['PRICE_PER_UNIT_AREA'].quantile(0.95)
# df_selected=df[(df['MAX_PRICE']<=price_limit) & (df['PRICE_PER_UNIT_AREA']<=ppu_limit)]
def remove_outlier(df,column):
    Q1=df[column].quantile(0.10)
    Q3=df[column].quantile(0.90)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    return df[(df[column]>=lower) & (df[column]<=upper)]
df_selected=remove_outlier(df_selected,'PRICE_PER_UNIT_AREA')
df_selected=remove_outlier(df_selected,'MAX_PRICE')
df_selected=remove_outlier(df_selected,'BEDROOM_NUM')
df_selected=remove_outlier(df_selected,'BATHROOM_NUM')
```

```python
df_selected=remove_outlier(df_selected,'BALCONY_NUM')
```

```python
df_selected=remove_outlier(df_selected,'TOTAL_FLOOR')
df_selected=remove_outlier(df_selected,'FLOOR_NUM')
# df_selected.info()
```

```python
# df_selected['AREA']=pd.to_numeric(df['AREA'])
df_selected=remove_outlier(df_selected,'AREA')
df_selected=remove_outlier(df_selected,'LOCATION_TARGET')
df_selected.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
X=df_selected[['BEDROOM_NUM','PRICE_PER_UNIT_AREA', 'FLOOR_NUM','LOCATION_TARGET','PROPERTY_TYPE_Independent/Builder Floor','PROPERTY_TYPE_Residential Apartment','AREA']]
y=df_selected['MAX_PRICE']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=47)
scaler = StandardScaler()
# X_train = scaler.fit_transform(X_train)
# X_test = scaler.transform(X_test)
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
diff = y_test - y_pred
plt.scatter(y_pred, diff,alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Predicted Values")
plt.ylabel("Diffrence in Prediction")
plt.title("Diffrence In Prediction Plot")
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor(random_state=47,max_depth=10)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
r2_score(y_test,y_pred)
```

```python
from sklearn.metrics import mean_absolute_percentage_error,mean_squared_error
mse=mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print('mean_squared_error:',mse)
print('root mean squared',rmse)
mape = mean_absolute_percentage_error(y_test, y_pred)
print("MAPE", mape * 100)
```

```python
diff = y_test - y_pred
plt.scatter(y_pred, diff,alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Predicted Values")
plt.ylabel("Diffrence In Prediction Price")
plt.title("Diffrence in Price prediction Plot")
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=300,max_depth=20)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print("R_2:", r2_score(y_test, rf_pred))
```

```python
diff=y_test-rf_pred
plt.scatter(y_test,diff,alpha=0.5)
plt.axhline(y=0,color='red',linestyle='--')
plt.ylabel('Prediction Diffrence')
plt.xlabel('Test Values')
```

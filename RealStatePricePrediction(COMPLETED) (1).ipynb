```python
import seaborn as sns
attributes=['number of bedrooms','number of bathrooms','living area','lot area','Price']
plt.figure(figsize=(12,8))
for i, col in enumerate(attributes, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(x=df[col], whis=1.5)
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
def remove_outlier(df,column):
    Q1=df[column].quantile(0.15)
    Q3=df[column].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    return df[(df[column]>=lower) & (df[column]<=upper)]
df=remove_outlier(df,'number of bedrooms')
df=remove_outlier(df,'number of bathrooms')
df=remove_outlier(df,'living area')
df=remove_outlier(df,'lot area')
df=remove_outlier(df,'Price')
df = df[df['Postal Code'] != 122071]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
X=df.drop(columns=['Price','Distance from the airport'],axis=1)
y=df['Price']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=47)
scaler = StandardScaler()
# X_train = scaler.fit_transform(X_train)
# X_test = scaler.transform(X_test)
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor(random_state=47,max_depth=10)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
r2_score(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=20,
    random_state=46
)

rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print("R_2:", r2_score(y_test, rf_pred))
```

```python
# server_fixed.py
from flask import Flask, request, jsonify
from flask_cors import CORS
import io
import base64
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import pandas as pd
import os

app = Flask(__name__)
CORS(app)
@app.route('/api/price', methods=['GET'])
def area_price():
    try:
        area = df.groupby('Postal Code', as_index=False)['Price'].mean()

        plt.figure(figsize=(12, 6))
        max_price = area['Price'].max()
        colors = ['red' if p == max_price else 'blue' for p in area['Price']]
        plt.bar(
            area['Postal Code'],
            area['Price'],
            width=0.5,
            alpha=0.8,
            color=colors
        )
        plt.ticklabel_format(style='plain', axis='both')

        plt.xticks(area['Postal Code'], rotation=90, fontsize=12)
        plt.xlabel("Postal Code", fontweight='bold', fontsize=12)
        plt.ylabel("Average Price", fontweight='bold', fontsize=12)
        # plt.gca().yaxis.set_major_formatter(mtick.FormatStrFormatter('%.1f'))
        plt.title("Average Price per Postal Code",fontweight='bold',fontsize=14)
        plt.tight_layout()

        img = io.BytesIO()
        plt.savefig(img, format='png', bbox_inches='tight')
        img.seek(0)
        plot_url = base64.b64encode(img.getvalue()).decode()
        plt.close()

        return jsonify({
            'success': True,
            'plotImage': f"data:image/png;base64,{plot_url}",
            'num_points': len(area)
        })
    except Exception as e:
        print("area_price error:", e)
        return jsonify({'success': False, 'error': str(e)}), 500


@app.route('/api/custom-plot', methods=['POST'])
def custom_stats():
    try:
        payload = request.get_json() or {}
        include = payload.get("include_postal_codes") 
        width = float(payload.get("width", 0.6))
        color = payload.get("color")
        include_list=[]
        include_list = include

        temp_df = df.copy()
        print(include_list)
        if include_list:
            temp_df = temp_df[temp_df['Postal Code'].isin(include_list)]
        if temp_df.empty:
            return jsonify({'success': False, 'error': "No data after filtering."}), 400

        area = temp_df.groupby('Postal Code', as_index=False)['Price'].mean().sort_values('Postal Code').reset_index(drop=True)

        x = list(range(len(area)))
        y = (area['Price'] / 100000).tolist()

        plt.figure(figsize=(max(8, len(x) * 0.25), 5), dpi=100,facecolor='#f5f5f5')
        plt.bar(x, y, width=width, color=color, alpha=0.85)

        plt.ticklabel_format(style='plain', axis='y')
        plt.title("Average Price per Postal Code In Lakhs", fontsize=12, fontweight='bold')
        plt.xlabel("Postal Code", fontsize=12, fontweight='bold')
        plt.ylabel("Average Price", fontsize=12, fontweight='bold')
        plt.xticks(x, area["Postal Code"], rotation=45, fontsize=10)
        plt.yticks(fontsize=10)
        plt.tight_layout()

        buf = io.BytesIO()
        plt.savefig(buf, format='png', bbox_inches='tight')
        buf.seek(0)
        img_b64 = base64.b64encode(buf.getvalue()).decode()
        plt.close()

        return jsonify({'success': True, 'plotImage': f"data:image/png;base64,{img_b64}", 'num_points': len(area)})

    except Exception as e:
        print("custom_plot error:", e)
        return jsonify({'success': False, 'error': str(e)}), 500


if __name__ == '__main__':
    # safe dev port
    app.run(host='0.0.0.0', port=8080, debug=False)
```
